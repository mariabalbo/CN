# 1. Sistemas lineares na forma matricial

a) Para o sistema:

$\begin{cases}2x+y=7\\ x-3y=-5\end{cases}$   

Temos:

Matriz de coeficientes ($A$): $\begin{pmatrix} 2 & 1 \\ 1 & -3 \end{pmatrix}$

Vetor de incógnitas ($x$): $\begin{pmatrix} x \\ y \end{pmatrix}$

Vetor independente ($b$): $\begin{pmatrix} 7 \\ -5 \end{pmatrix}$

b) Para o sistema:

$\begin{cases}x+2y-z=2\\ 3x-y+2z=9\\ 2x+y+z=7\end{cases}$

Temos:Matriz de coeficientes ($A$): $\begin{pmatrix} 1 & 2 & -1 \\ 3 & -1 & 2 \\ 2 & 1 & 1 \end{pmatrix}$

Vetor de incógnitas ($x$): $\begin{pmatrix} x \\ y \\ z \end{pmatrix}$

Vetor independente ($b$): $\begin{pmatrix} 2 \\ 9 \\ 7 \end{pmatrix}$


# 2. Método da eliminação de Gauss

Sistema:$\begin{cases}2x+y-z=8\\ -3x-y+2z=-11\\ -2x+y+2z=-3\end{cases}$   

Passo 1: Matriz Aumentada Inicial$\begin{pmatrix} 2 & 1 & -1 & \vert{} & 8 \\ -3 & -1 & 2 & \vert{} & -11 \\ -2 & 1 & 2 & \vert{} & -3 \end{pmatrix}$

Passo 2: Eliminação na primeira coluna

$L_2 \leftarrow L_2 - (\frac{-3}{2})L_1 \implies L_2 + 1.5 L_1$

$L_3 \leftarrow L_3 - (\frac{-2}{2})L_1 \implies L_3 + L_1$

Matriz resultante:$\begin{pmatrix} 2 & 1 & -1 & \vert{} & 8 \\ 0 & 0.5 & 0.5 & \vert{} & 1 \\ 0 & 2 & 1 & \vert{} & 5 \end{pmatrix}$

Passo 3: Eliminação na segunda coluna

$L_3 \leftarrow L_3 - (\frac{2}{0.5})L_2 \implies L_3 - 4 L_2$

Matriz Triangular resultante:$\begin{pmatrix} 2 & 1 & -1 & \vert{} & 8 \\ 0 & 0.5 & 0.5 & \vert{} & 1 \\ 0 & 0 & -1 & \vert{} & 1 \end{pmatrix}$

Passo 4: Substituição Regressiva

$-z = 1 \implies z = -1$

$0.5y + 0.5(-1) = 1 \implies 0.5y = 1.5 \implies y = 3$$2x + 3 - (-1) = 8 \implies 2x + 4 = 8 \implies 2x = 4 \implies x = 2$

Solução: $x = 2$, $y = 3$, $z = -1$.

# 3. Número de soluções

 a) $\begin{cases}x+y=3\\ 2x+2y=6\end{cases}$   

 Matriz aumentada: $\begin{pmatrix} 1 & 1 & \vert{} & 3 \\ 2 & 2 & \vert{} & 6 \end{pmatrix} \xrightarrow{L_2 \leftarrow L_2 - 2L_1} \begin{pmatrix} 1 & 1 & \vert{} & 3 \\ 0 & 0 & \vert{} & 0 \end{pmatrix}$

 Conclusão: Há infinitas soluções. A segunda equação anula-se (Sistema Possível e Indeterminado), restando apenas uma equação para duas variáveis.

 b) $\begin{cases}x+y=3\\ 2x+2y=7\end{cases}$   

 Matriz aumentada: $\begin{pmatrix} 1 & 1 & \vert{} & 3 \\ 2 & 2 & \vert{} & 7 \end{pmatrix} \xrightarrow{L_2 \leftarrow L_2 - 2L_1} \begin{pmatrix} 1 & 1 & \vert{} & 3 \\ 0 & 0 & \vert{} & 1 \end{pmatrix}$

 Conclusão: Nenhuma solução. A última linha traduz-se em $0 = 1$, o que é uma impossibilidade matemática (Sistema Impossível).

 c) $\begin{cases}x+y=3\\ 2x-y=0\end{cases}$   

 Matriz aumentada: $\begin{pmatrix} 1 & 1 & \vert{} & 3 \\ 2 & -1 & \vert{} & 0 \end{pmatrix} \xrightarrow{L_2 \leftarrow L_2 - 2L_1} \begin{pmatrix} 1 & 1 & \vert{} & 3 \\ 0 & -3 & \vert{} & -6 \end{pmatrix}$

 Conclusão: Solução única. O sistema é triangularizável sem zeros na diagonal principal, o que leva a $y=2$ e $x=1$ (Sistema Possível e Determinado).



# 4. Eliminação de Gauss

```python
import numpy as np

def gauss_elimination(A, b):
    Ab = np.hstack([A.astype(float), b.astype(float).reshape(-1, 1)])
    n = len(b)

    for i in range(n):
        for j in range(i+1, n):
            fator = Ab[j, i] / Ab[i, i]
            Ab[j, i:] -= fator * Ab[i, i:]

    x = np.zeros(n)
    for i in range(n-1, -1, -1):
        soma = np.sum(Ab[i, i+1:n] * x[i+1:n])
        x[i] = (Ab[i, -1] - soma) / Ab[i, i]

    return x

A_teste = np.array([[ 2,  1, -1],
                    [-3, -1,  2],
                    [-2,  1,  2]])
b_teste = np.array([ 8, -11, -3])

solucao_algoritmo = gauss_elimination(A_teste, b_teste)
solucao_numpy = np.linalg.solve(A_teste, b_teste)

print("Solução via nosso algoritmo de Eliminação de Gauss:", solucao_algoritmo)
print("Solução via numpy.linalg.solve:", solucao_numpy)
```

# 5. Modelo de balanço térmico

 $\begin{pmatrix} 4 & -1 & 0 \\ -1 & 4 & -1 \\ 0 & -1 & 3 \end{pmatrix} \begin{pmatrix} T_1 \\ T_2 \\ T_3 \end{pmatrix} = \begin{pmatrix} 15 \\ 10 \\ 10 \end{pmatrix}$   

 a) Resolução Manual:

 Matriz aumentada:

 $\begin{pmatrix} 4 & -1 & 0 & \vert{} & 15 \\ -1 & 4 & -1 & \vert{} & 10 \\ 0 & -1 & 3 & \vert{} & 10 \end{pmatrix}$

$L_2 \leftarrow L_2 + 0.25 L_1$

 $\begin{pmatrix} 4 & -1 & 0 & \vert{} & 15 \\ 0 & 3.75 & -1 & \vert{} & 13.75 \\ 0 & -1 & 3 & \vert{} & 10 \end{pmatrix}$

 $L_3 \leftarrow L_3 - (\frac{-1}{3.75})L_2 = L_3 + \frac{4}{15}L_2$

 $\begin{pmatrix} 4 & -1 & 0 & \vert{} & 15 \\ 0 & 3.75 & -1 & \vert{} & 13.75 \\ 0 & 0 & 2.733 & \vert{} & 13.666 \end{pmatrix}$

Substituição:

$\frac{41}{15} T_3 = \frac{41}{3} \implies T_3 = 5$$3.75 T_2 - 5 = 13.75 \implies 3.75 T_2 = 18.75 \implies T_2 = 5$$4 T_1 - 5 = 15 \implies 4 T_1 = 20 \implies T_1 = 5$

Solução: $T_1 = 5$, $T_2 = 5$, $T_3 = 5$.

```python
import numpy as np

A_term = np.array([[ 4, -1,  0],
                   [-1,  4, -1],
                   [ 0, -1,  3]])
b_term = np.array([15, 10, 10])

temperaturas = gauss_elimination(A_term, b_term)

print("Temperaturas calculadas:")
print(f"T1 = {temperaturas[0]:.1f}°")
print(f"T2 = {temperaturas[1]:.1f}°")
print(f"T3 = {temperaturas[2]:.1f}°")
```
